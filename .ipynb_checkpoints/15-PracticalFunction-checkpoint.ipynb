# Practical Functions

```python
def divideNumber(number):
    return number / 2
```

```python
divideNumber(20)
```

```python
myList = [3,5,7,10,20,30]
```

```python
myResultList = []
for num in myList:
    myResultList.append(divideNumber(num))
print(myResultList)
```

# MAP
Farklı hali yeni listeye ekler

```python
help(map)
```

```python
list(map(divideNumber,myList))
```

```python
def controlString(string):
    return "furkan" in string
```

```python
controlString("furkan taşdemir")
```

```python
myStringList = ["furkan","taş","gönül"]
```

```python
list(map(controlString,myStringList))
```

```python
#sadece furklan geçenleri tutmak istersem
```

# filter

```python
help(filter)
```

```python
list(filter(controlString,myStringList))
```

# LEGB -> local Enclosing Global Built-in

```python
#Global
myString = "furkan"

def myFunction():
    #enclosinh
    myString = "furkan 2"
    print(myString)
    
    def myFunction2():
        #Local
        myString = "furkan 3"
        print(myString)

    myFunction2()
```

```python
myString
```

```python
myFunction()
```
