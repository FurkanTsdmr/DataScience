```python
my_list = [20,30,40,50,60]
```

```python
#indexli yazdırma
for ix in range(len(my_list)):
    print(ix)
```

```python
for ix in range(len(my_list)):
    print(my_list[ix])
```

```python
#enumarete kolay yoldan yazdırma
```

```python
for element in enumerate (my_list):
    print(element)
```

```python
for (ix,value) in enumerate (my_list):
    print(value)
```
