### continue - break - pass

```python
my_list = [10,20,30,40,50,60,70]
```

```python
print("for loop started")
for num in my_list:
    print(num)
print("for loop ended")
```

```python
for number in my_list:
    print(number)
    if number == 40:
        print("yes")
    #buldugumuzda artık dur demek içni
    #continue break pass
        break #40 a erişince
```

```python
for number in my_list:
    if number == 40:
        continue
    print(number)
# 40 ı gelince direkt geçer
```

```python
for number in my_list:
    pass
```
