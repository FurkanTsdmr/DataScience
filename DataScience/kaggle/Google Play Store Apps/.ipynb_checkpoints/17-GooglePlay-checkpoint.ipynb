### Veri Tipi Dönüşümleri

````python
#diğer kolonlara bakalım mesela size'a bakalım
df_clean["Size"].value_counts()
#burada da bize int64 e uygun oldugunu söylüypr.
# 4   Size            10839 non-null  object fgakat object infodan bakınca
#detaylıca bakalım
````

````python
df_clean["Size"].unique()
#kb-mb değerleini alırız.
#nasıl alırız m yerine 3 0 koyup kb'a çevirirz.
````

````python
#m görüdpümüz yerde replace et neyi et 0 tane 0 
# k gördüğümüz yere hiçbir şey yazma sil kalsın öyle diyoruz.
df_clean["Size"] = df_clean["Size"].str.replace("M","000")
df_clean["Size"] = df_clean["Size"].str.replace("k","")
````

````python
#Varies with device gördüğün yere nan koy bir şey yazma diyoruz 
#ondan dolayı da .str ' yi kullanmıyoruz.
df_clean["Size"] = df_clean["Size"].replace("Varies with device",np.nan)
````

````python
* `Reviews` sütunundaki `3.0M` değerinin neden `int` dönüşümünü bozduğunu açıklayacağım.
* `Size` için `M → 000` yaklaşımının aslında **MB'yi KB'ye çevirmediğini**, burada dikkat edilmesi gereken birim dönüşümü olduğunu belirteceğim.
* `Installs` ve `Price` temizliğini `for` döngüsüyle açıklayacağım.
* `Last Updated` → `datetime` ve `.dt` kullanımını anlatacağım.
* `drop_duplicates(subset=["App"])` kararının ne anlama geldiğini açıklayacağım.
* `numeric_features` / `categorical_features` ayrımını göstereceğim.
* `plt.tight_layout` satırında `()` eksik olduğunu düzelteceğim.
* `kdeplot` ve `countplot` kullanımını tek tek açıklayacağım.
* En önemlisi, burada yaptığımız işlemin artık **gerçek bir veri temizleme + EDA pipeline'ına** dönüştüğünü vurgulayacağım.

# Data Science Yolculuğu #28: Google Play Dataset — Gerçek Bir Veri Setiyle Veri Analizi

Merhabalar tekrardan,

**Data Science Yolculuğu** serisinin 28. bölümüne hoş geldiniz.

Bir önceki bölümümüzde **Wine Quality** dataset'i üzerinden veri analizi yapmış ve farklı grafiklerle veriyi keşfetmeye çalışmıştık.

Bu bölümde ise biraz daha farklı bir dataset ile çalışacağız:

> **Google Play Store Apps**

Dataset'imizi Kaggle üzerinden inceleyebilirsiniz:

```text
https://www.kaggle.com/datasets/lava18/google-play-store-apps
```

Bu dataset içerisinde Google Play Store'daki uygulamalarla ilgili:

```text
App
Category
Rating
Reviews
Size
Installs
Type
Price
Content Rating
Genres
Last Updated
Current Ver
Android Ver
```

gibi bilgiler bulunuyor.

Bu bölümde sadece grafik çizmekle kalmayacağız.

Asıl önemli noktamız:

> **Gerçek bir veri seti geldiğinde onu nasıl temizleriz ve analize hazır hale getiririz?**

olacak.

Yani bu bölümde:

```text
Dataset'i Tanıma
        ↓
Veri Tiplerini İnceleme
        ↓
Eksik Verileri Kontrol Etme
        ↓
Hatalı / Uygunsuz Verileri Bulma
        ↓
Veri Tipi Dönüşümü
        ↓
String Temizliği
        ↓
Tarih Dönüşümü
        ↓
Duplicate Temizliği
        ↓
EDA
        ↓
Görselleştirme
```

şeklinde ilerleyeceğiz.

---

# Önce Kütüphanelerimizi Import Edelim

İlk olarak kullanacağımız kütüphaneleri import ediyoruz:

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
```

Burada:

```python
pandas
```

veri analizi ve DataFrame işlemleri için,

```python
numpy
```

sayısal işlemler ve `NaN` gibi değerler için,

```python
matplotlib
```

grafikler için,

```python
seaborn
```

ise daha gelişmiş ve kullanımı kolay istatistiksel grafikler için kullanılıyor.

---

# Dataset'imizi Okuyalım

Şimdi CSV dosyamızı Pandas ile okuyalım:

```python
df = pd.read_csv("googleplaystore.csv")
```

Artık dataset'imiz:

```text
df
```

değişkeninin içerisinde.

---

# İlk Olarak Dataset'e Bakalım

Bir dataset geldiğinde yapabileceğimiz ilk işlemlerden biri:

```python
df.head()
```

kullanmaktır.

Bu bize dataset'in ilk 5 satırını gösterir.

Neden ilk 5 satıra bakıyoruz?

Çünkü daha dataset'in tamamını analiz etmeden önce:

```text
Sütunlar neler?
Değerler nasıl görünüyor?
Veriler hangi formatta?
```

gibi sorulara hızlıca cevap almak istiyoruz.

---

# Sütunlarımız Neler?

Şimdi dataset içerisinde hangi sütunların bulunduğunu görelim:

```python
df.columns
```

Bu bize:

```text
App
Category
Rating
Reviews
Size
Installs
Type
Price
...
```

gibi sütun isimlerini verecektir.

Bu aşamada henüz herhangi bir işlem yapmıyoruz.

Sadece:

> **Elimizde ne var?**

onu anlamaya çalışıyoruz.

---

# Dataset'in Boyutu

Kaç satır ve kaç sütunumuz var?

```python
df.shape
```

Buradan:

```text
(satır sayısı, sütun sayısı)
```

şeklinde bir sonuç elde ederiz.

Örneğin:

```text
(10841, ...)
```

gibi bir değer görebiliriz.

Yani dataset'in yaklaşık 10 bin satırlık bir yapıya sahip olduğunu anlayabiliriz.

Bu noktada artık:

> "Elimde küçük bir tablo değil, ciddi miktarda kayıt var."

diyebiliriz.

---

# Şimdi En Önemli Kontrollerden Biri: info()

Şimdi:

```python
df.info()
```

çalıştıralım.

Burada bizim için çok önemli bir şey ortaya çıkıyor.

Dataset'te sayısal olması gereken birçok sütunun:

```text
object
```

olarak geldiğini görüyoruz.

Örneğin:

```text
Reviews
Size
Installs
Price
```

gibi sütunlar aslında sayısal değerler içeriyor.

Ama Pandas bunları:

```text
object
```

olarak görüyor.

Peki neden?

Çünkü verilerin içerisinde sadece sayı yok.

Örneğin:

```text
3.0M
```

gibi değerler var.

Başka bir sütunda:

```text
19M
14M
2.3M
```

gibi değerler bulunuyor.

Başka bir yerde:

```text
$4.99
```

gibi karakterler var.

Başka bir yerde:

```text
1,000,000+
```

gibi semboller bulunuyor.

Yani bilgisayarın gözünde bunlar doğrudan:

```text
1000000
```

değil.

Bunlar:

```text
string
```

olarak geliyor.

Burada artık şunu anlıyoruz:

> **Bu dataset'i analiz etmeden önce ciddi bir veri temizleme ve veri tipi dönüşümü yapmamız gerekiyor.**

---

# describe() Bize Ne Söylüyor?

Şimdi:

```python
df.describe()
```

çalıştıralım.

`describe()` varsayılan olarak sayısal sütunlar için istatistiksel özet verir.

Ancak burada dataset'in çoğu sütunu:

```text
object
```

olduğu için beklediğimiz kadar fazla sütun göremiyoruz.

Yani problem artık daha net:

> **Verilerimiz sayısal olmasına rağmen Pandas onları sayısal olarak tanımıyor.**

Bu problemi birazdan çözeceğiz.

---

# Eksik Verilerimiz Var mı?

Şimdi dataset'teki eksik değerleri kontrol edelim:

```python
df.isnull().sum()
```

Burada her sütunun kaç tane `null` değer içerdiğini görebiliriz.

Örneğin:

```text
Rating       ...
Size         ...
...
```

gibi sonuçlar çıkabilir.

Bu aşamada önemli olan:

> Eksik veri var mı?

sorusuna cevap bulmak.

Daha sonra bu eksik değerlerle ne yapacağımıza karar verebiliriz.

---

# Reviews Sütununa Biraz Daha Yakından Bakalım

Şimdi:

```python
df["Reviews"].value_counts()
```

çalıştıralım.

Burada ilginç bir durum görüyoruz.

`Reviews` sütunu:

```text
object
```

olmasına rağmen içerisindeki değerlerin büyük çoğunluğu sayısal görünüyor.

Örneğin:

```text
2
10
100
1000
...
```

gibi değerler var.

Bu durumda aklımıza hemen şu soru geliyor:

> "Bu sütunu neden integer yapmıyoruz?"

Deneyelim:

```python
df["Reviews"] = df["Reviews"].astype(int)
```

Fakat hata alıyoruz.

Örneğin:

```text
ValueError:
invalid literal for int() with base 10: '3.0M'
```

gibi bir hata görebiliriz.

Python aslında bize çok açık bir şey söylüyor:

> "Sen benden bütün sütunu integer yapmamı istiyorsun ama içerisinde `3.0M` diye bir değer var. Ben bunu doğrudan integer'a çeviremem."

---

# Problemli Değeri Bulalım

Şimdi `Reviews` içerisindeki değerlerin sayısal olup olmadığını kontrol edelim.

```python
df["Reviews"].str.isnumeric().sum()
```

Burada bize kaç tane değerin yalnızca sayısal karakterlerden oluştuğunu gösteriyor.

Dataset'te:

```text
10841
```

kayıt varsa ve bunun:

```text
10840
```

tanesi numeric çıkıyorsa geriye:

```text
1
```

problemli kayıt kalıyor.

Bu oldukça önemli bir bulgu.

---

# Sayısal Olmayan Değeri Bulalım

Şimdi:

```python
df[~df["Reviews"].str.isnumeric()]
```

yazalım.

Buradaki:

```python
~
```

operatörü boolean sonucu tersine çevirir.

Yani:

```text
True → False
False → True
```

hale gelir.

Normalde:

```python
df["Reviews"].str.isnumeric()
```

bize numeric olanların `True` değerini verir.

Biz ise:

```python
~
```

kullanarak:

> **Numeric olmayanları getir.**

diyoruz.

Sonuçta:

```text
3.0M
```

değerini buluyoruz.

---

# Neden Direkt Silmek Yerine Önce Kopya Alıyoruz?

Veri temizleme işlemlerinde orijinal dataset üzerinde doğrudan çalışmak yerine bir kopya oluşturmak daha güvenlidir.

```python
df_clean = df.copy()
```

Artık:

```text
df
```

orijinal verimiz.

```text
df_clean
```

ise üzerinde temizlik yaptığımız kopyamız.

Bu oldukça güzel bir alışkanlıktır.

Çünkü yaptığımız bir işlem sonucunda:

```text
"Yanlış bir şey yaptım."
```

dersek orijinal dataset elimizde kalır.

---

# Problemli Satırı Bulup Çıkaralım

Problemli değerin bulunduğu satırın index'ini belirledikten sonra:

```python
df_clean = df_clean.drop(df_clean.index[10472])
```

ile ilgili satırı çıkarıyoruz.

Ardından:

```python
df_clean.info()
```

ile kontrol ediyoruz.

Artık:

```text
10840
```

kayıt kaldığını görebiliriz.

Burada önemli bir nokta var:

> Gerçek projelerde bir satırı silmeden önce neden silindiğini mutlaka araştırmalıyız.

Çünkü:

```text
Problemli değer
```

her zaman:

```text
Hatalı kayıt
```

anlamına gelmez.

Biz bu örnekte veri temizleme pratiği yapıyoruz.

---

# Reviews Artık Integer Olabilir

Problemli değeri temizledikten sonra:

```python
df_clean["Reviews"] = df_clean["Reviews"].astype(int)
```

çalıştırabiliriz.

Ardından:

```python
df_clean.info()
```

diyerek tekrar kontrol edelim.

Artık:

```text
Reviews
int64
```

olarak görünmelidir.

Böylece ilk veri tipi dönüşümümüzü gerçekleştirdik.

---

# describe() Artık Daha Anlamlı

Şimdi:

```python
df_clean.describe()
```

çalıştıralım.

Artık `Reviews` sayısal olarak tanındığı için:

```text
count
mean
std
min
25%
50%
75%
max
```

gibi istatistikleri görebiliriz.

İşte veri tipi dönüşümünün önemi burada ortaya çıkıyor.

Bir sütun sayısal olmasına rağmen:

```text
object
```

olarak kalırsa matematiksel analizlerimizi doğrudan yapamayız.

---

# Şimdi Size Geçelim

Bir sonraki problemli sütunumuz:

```text
Size
```

Önce değerlerin dağılımına bakalım:

```python
df_clean["Size"].value_counts()
```

Ardından:

```python
df_clean["Size"].unique()
```

ile hangi farklı değerlerin bulunduğuna bakalım.

Burada:

```text
19M
14M
8.7M
2.3M
...
```

gibi değerlerin yanında:

```text
512k
1.4M
```

gibi farklı birimlerle ifade edilmiş değerler görebiliriz.

Ayrıca:

```text
Varies with device
```

gibi doğrudan sayısal olmayan bir değer de bulunuyor.

Dolayısıyla bunu da temizlememiz gerekiyor.

---

# Size İçerisindeki M ve k Problemi

Burada dikkat etmemiz gereken önemli bir konu var.

Dataset'teki:

```text
M
```

megabyte,

```text
k
```

ise kilobyte anlamına geliyor.

Biz bunları doğrudan:

```text
M → 000
k → ""
```

şeklinde değiştirirsek aslında **birim dönüşümünü matematiksel olarak tam yapmış olmayız.**

Örneğin:

```text
1M
```

değerini:

```text
1000
```

yapmak onu yaklaşık olarak:

```text
1000 KB
```

şeklinde ifade etmeye dayanır.

Ama:

```text
1.4M
```

değerini:

```text
1.400 KB
```

olarak yorumlamamız gerekir.

Burada temel amaç:

> Farklı birimlerdeki değerleri aynı birime getirerek karşılaştırılabilir hale getirmek.

---

# String İşlemleri ile Temizlik

Örneğimizde önce `M` ve `k` karakterlerini temizleyebiliriz.

```python
df_clean["Size"] = df_clean["Size"].str.replace("M", "000")
df_clean["Size"] = df_clean["Size"].str.replace("k", "")
```

Burada:

```python
.str
```

Pandas'ın string değerler üzerinde toplu işlem yapmamızı sağlayan yapısıdır.

Örneğin:

```python
.str.replace()
```

ile belirli karakterleri değiştirebiliriz.

---

# Varies with device Problemi

Şimdi:

```python
df_clean["Size"].unique()
```

ile tekrar bakalım.

Karşımıza:

```text
Varies with device
```

çıkacaktır.

Bu değer sayısal olmadığı için doğrudan `float` dönüşümü yapamayız.

Burada:

> "Bu değer bilinmiyor."

anlamında `NaN` kullanabiliriz.

```python
df_clean["Size"] = df_clean["Size"].replace(
    "Varies with device",
    np.nan
)
```

Artık bu değer:

```text
NaN
```

oldu.

Yani:

> Size bilgisi mevcut değil.

şeklinde temsil ediliyor.

---

# Size'ı Float'a Çevirelim

Artık sayısal olmayan:

```text
Varies with device
```

değerini temizledik.

Şimdi:

```python
df_clean["Size"] = df_clean["Size"].astype(float)
```

diyebiliriz.

Neden `float`?

Çünkü:

```text
8.7M
14.2M
1.4M
```

gibi ondalıklı değerler bulunabiliyor.

Dolayısıyla:

```text
int
```

yerine:

```text
float
```

kullanıyoruz.

---

# Installs Sütununa Geçelim

Şimdi:

```python
df_clean["Installs"].value_counts()
```

ve:

```python
df_clean["Installs"].unique()
```

ile değerleri inceleyelim.

Burada:

```text
1,000+
10,000+
100,000+
1,000,000+
```

gibi değerler görüyoruz.

Problemlerimiz:

```text
+
,
```

karakterleri.

Bunlar sayısal değerlerin içerisinde bulunduğu için Pandas sütunu:

```text
object
```

olarak algılıyor.

---

# Price Sütununda da Benzer Problem Var

Şimdi:

```python
df_clean["Price"].value_counts()
```

ile Price sütununa bakalım.

Burada:

```text
$4.99
$2.99
$0.99
```

gibi değerler görebiliriz.

Buradaki problem:

```text
$
```

karakteri.

Dolayısıyla:

```text
Installs
Price
```

sütunlarında string temizliği yapmamız gerekiyor.

---

# Aynı İşlemi Tekrar Tekrar Yapmayalım

Şöyle tek tek yazmak yerine:

```python
df_clean["Installs"] = ...
df_clean["Price"] = ...
```

bir döngü oluşturabiliriz.

Önce temizlenecek karakterleri belirleyelim:

```python
chars_to_remove = ["+", ",", "$"]
```

Sonra bu karakterlerin bulunduğu sütunları:

```python
cols_to_remove = ["Installs", "Price"]
```

şeklinde tanımlayalım.

Artık Python'a:

> "Bu karakterleri bu sütunların tamamından temizle."

diyebiliriz.

---

# İç İçe For Döngüsü

```python
for item in chars_to_remove:
    for cols in cols_to_remove:
        df_clean[cols] = df_clean[cols].str.replace(item, "")
```

Burada iki tane `for` var.

İlk döngü:

```python
for item in chars_to_remove:
```

şunları tek tek dolaşır:

```text
+
,
$
```

İkinci döngü:

```python
for cols in cols_to_remove:
```

şunları dolaşır:

```text
Installs
Price
```

Sonuç olarak:

```text
+
↓
Installs
Price

,
↓
Installs
Price

$
↓
Installs
Price
```

şeklinde bütün kombinasyonları temizlemiş oluyoruz.

---

# Artık Price'ı Float Yapabiliriz

Temizleme işlemi bittikten sonra:

```python
df_clean["Price"] = df_clean["Price"].astype(float)
```

diyoruz.

Artık:

```text
Price
```

sayısal bir sütun.

---

# Installs'ı Integer'a Çevirelim

Aynı şekilde:

```python
df_clean["Installs"] = df_clean["Installs"].astype(int)
```

diyebiliriz.

Artık:

```text
Installs
```

da sayısal hale geldi.

---

# Şimdi Dataset'i Biraz Daha Genel İnceleyelim

Temizleme işlemlerimizden sonra:

```python
sns.pairplot(df_clean)
```

ile dataset'in değişkenlerini genel olarak karşılaştırmayı deneyebiliriz.

Burada Pairplot bize sayısal değişkenlerin ikili ilişkilerini gösterir.

Ancak burada önemli bir detay var:

Eğer dataset içerisinde:

```text
Id
```

gibi sadece kayıt tanımlamak için kullanılan bir sütun varsa bunu analize dahil etmek genellikle anlamlı değildir.

Çünkü:

> ID bir ölçüm değildir.

Bu nedenle daha kontrollü bir Pairplot için analiz dışı ID sütunlarını çıkarmak daha doğru olabilir.

Örneğin:

```python
sns.pairplot(
    df_clean.drop(columns=["Id"], errors="ignore")
)
```

şeklinde kullanabiliriz.

---

# Last Updated Bir String

Şimdi:

```python
df_clean["Last Updated"]
```

ile tarih sütunumuza bakalım.

Ardından:

```python
df_clean["Last Updated"].unique()
```

ile değerleri inceleyebiliriz.

Örneğin:

```text
January 7, 2018
August 3, 2017
...
```

gibi değerler görebiliriz.

Buradaki problem:

> Bunlar tarih gibi görünse de Pandas açısından henüz gerçek bir tarih veri tipi olmayabilir.

---

# Tarihi Datetime'a Çevirelim

Bunun için:

```python
df_clean["Last Updated"] = pd.to_datetime(
    df_clean["Last Updated"]
)
```

kullanıyoruz.

Artık Pandas bu sütunu tarih olarak tanıyabiliyor.

Bu çok önemli.

Çünkü artık:

```text
Gün
Ay
Yıl
```

gibi parçaları kolayca ayırabiliriz.

---

# Tarihten Gün, Ay ve Yıl Çıkaralım

Gün:

```python
df_clean["Day"] = df_clean["Last Updated"].dt.day
```

Ay:

```python
df_clean["Month"] = df_clean["Last Updated"].dt.month
```

Yıl:

```python
df_clean["Year"] = df_clean["Last Updated"].dt.year
```

Buradaki:

```python
.dt
```

Pandas'ın datetime özelliklerine erişmemizi sağlar.

Örneğin:

```python
.dt.day
```

→ gün

```python
.dt.month
```

→ ay

```python
.dt.year
```

→ yıl

bilgisini verir.

---

# Duplicate Uygulamalar Var mı?

Şimdi önemli bir kontrol daha yapalım.

Dataset'te aynı uygulama birden fazla kez bulunuyor olabilir.

Bunu:

```python
df_clean[df_clean.duplicated("App")].shape
```

ile kontrol edebiliriz.

Burada:

```python
duplicated("App")
```

şunu soruyor:

> "App sütununa göre daha önce aynı uygulama var mı?"

---

# Duplicate Uygulamaları Temizleyelim

Eğer aynı uygulamadan birden fazla kayıt varsa:

```python
df_clean = df_clean.drop_duplicates(
    subset=["App"],
    keep="first"
)
```

kullanabiliriz.

Burada:

```python
subset=["App"]
```

ile duplicate kontrolünü sadece `App` sütununa göre yapıyoruz.

```python
keep="first"
```

ise:

> Aynı uygulamanın ilk kaydını tut, sonraki tekrarları sil.

anlamına geliyor.

Bu işlemden sonra dataset'in satır sayısı azalacaktır.

Örneğin yaklaşık:

```text
10840
```

satırdan:

```text
9659
```

satıra düşebiliriz.

Burada artık dataset'in daha temiz bir versiyonuna sahibiz.

---

# Sayısal ve Kategorik Sütunları Ayıralım

Şimdi analizimizi kolaylaştırmak için sütunları iki gruba ayıralım:

```text
Numeric
Categorical
```

Bunu şöyle yapabiliriz:

```python
numeric_features = [
    feature
    for feature in df_clean.columns
    if df_clean[feature].dtype != "O"
]
```

Burada:

```python
dtype != "O"
```

diyerek object olmayan sütunları alıyoruz.

Bunlar bizim sayısal veya datetime gibi sayısal olmayan ama object de olmayan değişkenlerimizi içerebilir.

Kategorik sütunlar için:

```python
categorical_features = [
    feature
    for feature in df_clean.columns
    if df_clean[feature].dtype == "O"
]
```

kullanıyoruz.

Burada:

```text
O
```

Pandas'taki `object` dtype'ı temsil ediyor.

---

# Sayısal Değişkenlerin Dağılımını Görelim

Şimdi elimizdeki sayısal özelliklerin dağılımlarını tek bir Figure üzerinde göstermek istiyoruz.

Önce:

```python
plt.figure(figsize=(15,10))
```

ile Figure boyutumuzu belirliyoruz.

Sonra:

```python
for i in range(0, len(numeric_features)):
```

ile sayısal sütunların tamamını dolaşıyoruz.

Her bir sütun için:

```python
plt.subplot(5,3,i+1)
```

ile grafik alanı oluşturuyoruz.

Ardından:

```python
sns.kdeplot(
    x=df_clean[numeric_features[i]],
    fill=True
)
```

ile ilgili sütunun KDE dağılımını çiziyoruz.

---

# Neden For Döngüsü Kullanıyoruz?

Diyelim ki elimizde:

```text
Rating
Reviews
Size
Installs
Price
Day
Month
Year
```

gibi 8 sayısal sütun var.

Bunların her biri için ayrı ayrı:

```python
sns.kdeplot(...)
```

yazmak yerine:

```python
for
```

kullanıyoruz.

Python'a:

> "Sayısal sütunların hepsini dolaş ve her biri için bir grafik oluştur."

demiş oluyoruz.

Bu, kodumuzu çok daha kısa ve tekrar kullanılabilir hale getiriyor.

---

# Burada Küçük Bir Kod Detayı

Kodumuzda:

```python
plt.tight_layout
```

şeklinde yazarsak fonksiyonu çağırmış olmayız.

Doğru kullanım:

```python
plt.tight_layout()
```

olmalı.

Çünkü:

```python
tight_layout
```

fonksiyonun kendisi,

```python
tight_layout()
```

ise fonksiyonun çalıştırılmasıdır.

Bu Python'da çok sık karşılaşacağımız bir ayrımdır.

---

# Kategorik Değişkenleri İnceleyelim

Şimdi:

```text
Type
Content Rating
```

gibi kategorik değişkenlerimize bakalım.

Önce:

```python
category = [
    "Type",
    "Content Rating"
]
```

şeklinde sütunlarımızı belirliyoruz.

Sonra:

```python
plt.figure(figsize=(15,5))
```

ile Figure oluşturuyoruz.

Ardından:

```python
for i in range(0, len(category)):
```

ile kategorilerimizi dolaşıyoruz.

---

# Countplot ile Kategorileri Gösterelim

Her kategori için:

```python
plt.subplot(1,2,i+1)
```

ile grafik alanı oluşturuyoruz.

Sonra:

```python
sns.countplot(
    x=df_clean[category[i]]
)
```

kullanıyoruz.

`countplot` bize:

> Her kategoriden kaç tane olduğunu

gösterir.

Örneğin `Type` için:

```text
Free
Paid
```

kaç tane uygulama olduğunu görebiliriz.

`Content Rating` için de:

```text
Everyone
Teen
Mature 17+
...
```

gibi kategorilerin dağılımını inceleyebiliriz.

---

# Burada Ne Öğrendik?

Artık dataset'i sadece:

```text
Okuyup
```

grafik çizmedik.

Gerçek bir veri temizleme sürecinden geçirdik.

Başlangıçta:

```text
Reviews → object
Size → object
Installs → object
Price → object
```

gibi problemli sütunlarımız vardı.

Bunların içerisinde:

```text
M
k
+
,
$
```

gibi karakterler bulunuyordu.

Ayrıca:

```text
3.0M
Varies with device
```

gibi doğrudan sayıya çevrilemeyen değerlerle karşılaştık.

Bunları temizleyip uygun veri tiplerine dönüştürdük.

---

# Veri Temizleme Sürecimiz

Yaptığımız işlemleri bir akış halinde düşünelim:

```text
Google Play Dataset
        ↓
       head()
        ↓
      columns
        ↓
       shape
        ↓
       info()
        ↓
   Eksik değer kontrolü
        ↓
 Reviews temizliği
        ↓
 Reviews → int
        ↓
 Size temizliği
        ↓
 Size → float
        ↓
 Installs temizliği
        ↓
 Installs → int
        ↓
 Price temizliği
        ↓
 Price → float
        ↓
 Tarih → datetime
        ↓
 Duplicate temizliği
        ↓
 Numeric / Categorical ayrımı
        ↓
      Görselleştirme
```

Bu aslında gerçek hayattaki veri analizinin önemli bir bölümünü oluşturuyor.

---

# Burada Çok Önemli Bir Şey Fark Edelim

Bir dataset'i Kaggle'dan indirip:

```python
sns.pairplot(df)
```

yazmak çok kolay.

Ama gerçek hayatta veriler her zaman:

```text
10
20
30
40
```

şeklinde tertemiz gelmiyor.

Karşımıza:

```text
10M
2.5M
1,000+
$4.99
Unknown
Varies with device
```

gibi değerler çıkabiliyor.

Data Scientist'in işi sadece grafik çizmek değil.

Önce:

> **Bu veri gerçekten analiz edilebilir durumda mı?**

sorusunu sormak gerekiyor.

---

# Veri Tipi Neden Bu Kadar Önemli?

Şöyle düşünelim.

Elimizde:

```text
"100"
"200"
"300"
```

olsun.

Bunlar string ise bilgisayar bunları metin olarak görebilir.

Ama biz:

```text
100 + 200 + 300
```

hesaplamak istiyoruz.

Bu durumda verileri:

```text
int
```

veya:

```text
float
```

haline getirmemiz gerekir.

Yani:

```text
Görünen değer
≠
Bilgisayarın onu nasıl algıladığı
```

olabilir.

Data Science'ta bu ayrımı anlamak oldukça önemlidir.

---

# EDA'ya Artık Hazırız

Veri temizleme işlemlerimizi yaptıktan sonra artık dataset üzerinde daha sağlıklı analizler yapabiliriz.

Örneğin:

```python
df_clean.describe()
```

ile sayısal değişkenlerin istatistiklerini inceleyebiliriz.

```python
sns.pairplot(df_clean)
```

ile değişkenlerin ilişkilerine bakabiliriz.

```python
sns.kdeplot(...)
```

ile dağılımları inceleyebiliriz.

```python
sns.countplot(...)
```

ile kategorilerin frekanslarını görebiliriz.

Artık grafiklerin bize verdiği bilgiler daha anlamlı hale geliyor.

Çünkü önce:

> **Verinin kendisini temizledik.**

---

# Data Science'ta Bence En Önemli Alışkanlıklardan Biri

Bir dataset gördüğümüzde hemen:

```text
Grafik çiz.
Model kur.
Tahmin yap.
```

dememeliyiz.

Önce:

```text
Bu veri nereden geliyor?
```

```text
Sütunlar ne anlama geliyor?
```

```text
Veri tipi doğru mu?
```

```text
Eksik değer var mı?
```

```text
Duplicate var mı?
```

```text
String içerisinde sayı var mı?
```

```text
Birimler aynı mı?
```

```text
Tarih gerçekten tarih olarak mı tutuluyor?
```

gibi soruları sormalıyız.

Çünkü:

> **Kirli veriyle yapılan çok güzel bir analiz, yine de yanlış bir analiz olabilir.**

---

# Son Bir Kez Mantığı Oturtalım

Bu bölümde Google Play Store dataset'i üzerinden gerçek bir veri temizleme ve keşif süreci gerçekleştirdik.

Önce:

```text
Dataset'i tanıdık.
```

Sonra:

```text
Eksik değerleri kontrol ettik.
```

Ardından:

```text
Yanlış / uygun olmayan veri tiplerini bulduk.
```

Sonra:

```text
Reviews
Size
Installs
Price
```

sütunlarını temizledik.

Ardından:

```text
String → Integer
String → Float
String → Datetime
```

gibi dönüşümler gerçekleştirdik.

Sonrasında:

```text
Duplicate
```

kayıtları kontrol edip temizledik.

En sonunda da:

```text
Numeric Features
Categorical Features
```

olarak değişkenlerimizi ayırıp:

```text
KDE
Countplot
Pairplot
```

gibi grafiklerle verimizi keşfetmeye başladık.

Bunu kafamızda şöyle tutabiliriz:

```text
             GERÇEK DATASET
                    ↓
              Önce Tanı
                    ↓
             Veri Kalitesi
                    ↓
       ┌────────────┼────────────┐
       ↓            ↓            ↓
    Missing      Duplicate    Data Type
       ↓            ↓            ↓
                  Temizle
                    ↓
              Dönüştür
                    ↓
              Analiz Et
                    ↓
             Görselleştir
                    ↓
                  EDA
```

Ve bu bölümde aslında çok önemli bir noktaya geldik:

> **Data Science, sadece Python ile grafik çizmek değildir.**

Asıl mesele:

> **Dağınık ve ham veriyi anlayıp, temizleyip, analiz edilebilir hale getirmek ve sonrasında doğru soruları sorabilmektir.**

Google Play dataset'i bize bunu oldukça güzel gösteriyor.

Çünkü ilk baktığımızda:

```text
Reviews → object
Size → object
Installs → object
Price → object
```

gibi görünen bir dataset vardı.

Biraz uğraştığımızda ise:

```text
Reviews → int
Size → float
Installs → int
Price → float
Last Updated → datetime
```

şeklinde analiz edilebilir bir veri yapısına dönüştürdük.

İşte veri temizleme tam olarak burada başlıyor.

Bir sonraki aşamada ise artık bu temizlediğimiz veriye daha anlamlı sorular sorabiliriz:

```text
En çok indirilen uygulamalar hangileri?
Hangi kategoriler daha fazla uygulamaya sahip?
Ücretli ve ücretsiz uygulamalar arasında fark var mı?
Rating ile Reviews arasında ilişki var mı?
Uygulama büyüklüğü ile indirme sayısı arasında ilişki var mı?
Hangi kategoriler daha yüksek puanlara sahip?
```

Yani artık:

> **"Veri elimizde ne var?"**

aşamasından,

> **"Bu veriden ne öğrenebiliriz?"**

aşamasına geçiyoruz.

İşte Data Science yolculuğunda asıl eğlenceli kısım da burada başlıyor.
````
