# 12 - Boxplot

```python
import numpy as np
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# box plot
# median,quartile(25%,50%,75%),mix-max 1.5 x IQR ,outlier
```

```python
data = np.array([5,7,9,18,24,33,27,37,33,35,67,46,47,50,55,60,100])
plt.figure(figsize=(6,5))
sns.boxplot(y=data)
plt.title("Box Plot")
plt.ylabel("Data Value")
#plt.grid(True)
plt.show()
#medyanı mavi alanın oratasında çizdiği çizgide gösteriyor.
#genişliğin büyüdüğü yerlerde datanın hangi alanlarda genişlediğini gösteriypr.
#üst ve alt çizgileri nereye kadar çıkıp indiiğini gmsteriyor.
#Üst noktada ki daire ise çıkıntısıdır o çıkıntıda 100 dür.outlierdır
#
```

```python
data = np.array([5,7,9,15,20,22,25,30,32,35,37,40,50,55,60,100])
plt.figure(figsize=(6,5))
sns.boxplot(y=data)
plt.title("Box Plot")
plt.ylabel("Data Value")
#plt.grid(True)
plt.show()
#bu sefer atynı tekrar eden değerler oldugu zaman da bakın ne olacak
#medyan daha yukarda ve değerlein birleştiği noktada bir laanda 
#yer kaplamış
```

# Seaborn Data Set

```python
#kaggle gibi sitelerden indirebiliyoruz ayrıyetten 
#https://github.com/mwaskom/seaborn-data bu arkadaşta datasetleri 
#hazır şekilde vermiş.
```

```python
df = sns.load_dataset("titanic")
```

```python
df
```

```python
plt.figure(figsize=(13,8))
plt.subplot(1,2,1)
sns.boxplot(x="class",y="age",data=df)
plt.title("Age by Class")
plt.xlabel("Class")
plt.ylabel("Age")
#1 satır olsun 2 sütun olsun şu anda 1.çiziyorum
plt.subplot(1,2,2)
sns.boxplot(x="class",y="fare",data=df)
plt.title("Fare by Class")
plt.xlabel("Class")
plt.ylabel("Fare")
plt.tight_layout()
plt.show()
#ilk grafikte yaşlara göre bakıyoruz ve şunu diyebiliriz.
#yaş arttıkca class değeri 1.2.3.sınıf değerleri artmış,
#yani yaş yükseldikçe zenginlik artmış, medtanları 36 civarında
#3.sınıfta olup outlier olanları 75 yaşta olanlarda var 2
#2.sınıfta outlıer cok altta olanlarda var 
#1.sınıflar içerisinde daha dengeli var outlier var ama daha dengeli.
#ödenmiş parada fare by classta first classta daha dengeli bir dağışım var fakat büyük bir outlier var
#medyanı 50 lerde civarı.
#2.ve3. sınıflar daha düşük ödemiş 3.sınıfta outlıer var fakat 
#b 3.sınıf first classa'a binmek için daha yüksek ödemiş.
```
