```python
sns.set_style(style="dark")
sns.displot(x="fixed acidity",data=df)

plt.title("Acid vs PH")
plt.xlabel("Acid")
plt.ylabel("PH")
plt.show()
```

```python
sns.set_style(style="dark")
sns.displot(x="fixed acidity",data=df,kind="kde")

plt.title("Acid vs PH")
plt.xlabel("Acid")
plt.ylabel("PH")
plt.show()
```

```python
df.columns
```

```python
sns.catplot(x="chlorides",y="density",data=df)
plt.title("Chlorides vs Destinit")
plt.show()
```

```python
sns.barplot(x="citric acid",y="alcohol",data=df)
plt.title("Citric Acid vs Alcohol")
plt.show()
```

```python
plt.figure(figsize=(10,7))
sns.heatmap(df.corr(numeric_only=True),annot=True)
```

```python
df.head()
```

```python
#tum colonların dağılımlarını almak
columns = df.columns
(fig,ax)=plt.subplots(4,4,figsize=(16,15))
#print(ax.flatten())
ax = ax.flatten()

for i,column in enumerate(columns):
    #print(i,column)
    #her bir grafik için bir tane colona atama yapıcaz
    sns.kdeplot(
        data=df,
        x= column,
        hue=df.quality,
        ax=ax[i]
    )
    ax[i].set_title(f"{column}Distribution")
    ax[i].set_xlabel(None)
#range olusturuyoruz, boş olan tabloları kapatıyor
for i in range(i+1,len(ax)):
    ax[i].axis("off")
```
