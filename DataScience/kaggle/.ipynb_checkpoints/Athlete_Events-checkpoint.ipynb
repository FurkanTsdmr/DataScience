```python
plt.scatter("Height","Weight",data=df)
plt.xlabel("Height")
plt.ylabel("Weight")
plt.title("Athelte Weight vs Height")
```

```python
sns.set_style("whitegrid")
sns.scatterplot(x="Height",y="Weight", data=df)
plt.xlabel("Height of Atheletes")
plt.ylabel("Weight of Atheletes")
plt.title("Atheletes Weight vs Height")
plt.show()
```

```python
sns.set_style("dark")
sns.scatterplot(x="Height",y="Weight",hue="Sex", data=df)
plt.xlabel("Height of Atheletes")
plt.ylabel("Weight of Atheletes")
plt.title("Atheletes Weight vs Height")
plt.show()

#Veri setinizi kaydettiğiniz değişken adı neyse, 
#data= parametresine o ismi vermeniz gerekir (örneğin veri setinizin adı df ise):
```

```python
sns.set_style("whitegrid")
sns.scatterplot(x="Height",y="Weight",hue="Medal", data=df)
plt.xlabel("Height of Atheletes")
plt.ylabel("Weight of Atheletes")
plt.title("Atheletes Weight vs Height")
plt.show()
```

```python
sns.set_style("dark")
sns.scatterplot(x="Height",y="Weight",hue="Sex",style="Medal", data=df)
plt.xlabel("Height of Atheletes")
plt.ylabel("Weight of Atheletes")
plt.title("Atheletes Weight vs Height")
plt.show()
```

```python
sns.set_style("dark")
sns.scatterplot(x="Height",y="Weight",hue="Sex",style="Medal",size="Age", data=df)
plt.xlabel("Height of Atheletes")
plt.ylabel("Weight of Atheletes")
plt.title("Atheletes Weight vs Height")
plt.show()
```

```python
sns.set_style("whitegrid")
sns.lineplot(x="Height",y="Weight",data=df,hue="Sex")
plt.xlabel("Height of Atheletes")
plt.ylabel("Weight of Atheletes")
plt.title("Atheletes Weight vs Height")
plt.show()
```

```python
sns.set_style("white")
sns.displot(x="Height",hue="Sex",data=df)
plt.ylabel("Frequency")
plt.title("Athletes Height Distribution")
plt.show()
```

```python
sns.set_style("white")
sns.displot(x="Height",hue="Sex",data=df,kind="kde")
plt.ylabel("Frequency")
plt.title("Athletes Height Distribution")
plt.show()
```

```python
sns.barplot(x="Medal",y="Height",hue="Sex",data=df)
plt.title("Medals by Height")
plt.show()
```

```python
sns.catplot(x="Medal",y="Weight",hue="Sex",data=df,col="Season")
plt.title("Season medal")
plt.show()
```

```python
sns.catplot(x="Medal",y="Weight",hue="Sex",data=df,col="Season",kind="bar")
plt.title("Season medal")
plt.show()
```
