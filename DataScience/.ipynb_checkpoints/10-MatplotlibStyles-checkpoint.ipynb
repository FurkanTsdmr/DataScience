# 10-MatplotlibStyles

```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
data1 = np.linspace(0,10,20)
```

```python
data1
```

```python
data2 = data1 ** 2
```

```python
plt.subplots()
```

```python
my_fig, my_axes = plt.subplots()
```

```python
type(my_fig)
```

```python
type(my_axes)
```

```python
my_fig, my_axes = plt.subplots()
my_axes.plot(data1,data2,color="#1E0703",alpha=0.5)
my_axes.plot(data2,data1,color="#53CA0A")
plt.show()
```

```python
(new_fig,new_axes) = plt.subplots()

new_axes.plot(data1,data1 + 2,color="blue",alpha=0.7,linewidth=1.0)
new_axes.plot(data1,data1 + 3,color="blue",alpha=0.7,linewidth=6.0)

new_axes.plot(data1,data1 + 4,color="blue",alpha=0.7,linewidth=3.0,linestyle="--")
new_axes.plot(data1,data1 + 7,color="red",alpha=0.7,linewidth=1.0)
new_axes.plot(data1,data1 + 8,color="#000000",alpha=0.7,linewidth=1.0,marker="o",markersize=8,markerfacecolor="green")
new_axes.plot(data1,data1 + 8,color="#000000",alpha=0.7,linewidth=1.0,marker="o",markersize=4,markerfacecolor="green")


plt.show()
```

```python
#scatter
```

```python
plt.scatter(data1,data2)
plt.show()
```
