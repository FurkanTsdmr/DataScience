9 - IntroToMatplotib

```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

```python
#daha güzel çalışma görsellerde problem olmaması için
#lab için geçerli sadece burası
%matplotlib inline
```

```python
age_list = [10,20,30,30,30,40,50,60,70,75]
```

```python
weight_list = [20,60,80,85,86,87,70,90,95,90]
```

```python
plt.plot(age_list,weight_list,"b") #y-> yellow,g,b,r
plt.xlabel("Age")
plt.ylabel("Weight")
plt.title("Age vs Weight")
plt.show()
```

```python
#numpy
np_age_list = np.array(age_list)
```

```python
np_weight_list = np.array(weight_list)
```

```python
plt.plot(np_age_list,np_weight_list,"g")
plt.title("Numpy Age vs Weight")
plt.xlabel("Age")
plt.ylabel("Weight")
plt.show()
```

```python
numpy_arr1=np.linspace(0,10,20)#0 dan 10 a kadar 20 tane sayı dön
#bunları dönüyor ve aralarını eşit tutuyor
```

```python
numpy_arr1
```

```python
numpy_arr2 = numpy_arr1 ** 3
```

```python
numpy_arr2
```

```python
plt.plot(numpy_arr1,numpy_arr2)
plt.show()
```

```python
plt.plot(numpy_arr1,numpy_arr2,"r*-")
plt.show()
```

```python
#Matplotlib Eksenler
```

```python
plt.plot(numpy_arr2,numpy_arr1,"r*-")
plt.show()
```

```python
plt.subplot(1,2,1)
#1 row 2 col 1.grafiği çiziyorum demek
plt.plot(numpy_arr1,numpy_arr2,"r*-")
plt.xlabel("age")
plt.ylabel("weight")
plt.title("*- Age vs Weight")
plt.subplot(1,2,2)
plt.plot(numpy_arr2,numpy_arr1,"g--")
plt.xlabel("Weight")
plt.ylabel("Age")
plt.title("-- Weight vs Age")
plt.show()
```

```python
#kendimnize özel yapıtk
my_figure = plt.figure()
figure_axes= my_figure.add_axes([0.1,0.1,0.5,0.5])
#([0.1,0.1,0.3,0.3]) 0-1 arası solda alttan küçültme büyümr
#ilk 0.3 0.7 yaptıgımızda genişlil
#en sagdaki yükseklil
#0.1 0.1 ise soldan ve alttan boşluk anlamına geliyor
figure_axes.plot(numpy_arr1,numpy_arr2,"g")
figure_axes.set_xlabel("X Axis")
figure_axes.set_ylabel("Y Label")
figure_axes.set_title("Graph Title")
plt.show()
```

```python
new_fig = plt.figure(dpi=200)

new_axes = new_fig.add_axes([0.1,0.1,0.9,0.9])

new_axes.plot(numpy_arr1, numpy_arr1 ** 2, label= 'numpy array **2')
new_axes.plot(numpy_arr1, numpy_arr1 ** 3, label= 'numpy array **3')
#küçük bir bilgikendirme veriyor loc nerde duracagını belirtiyoruz
new_axes.legend(loc=1)
plt.show()
```

```python
age_list = [12, 18, 25, 31, 38, 45, 52, 60, 68, 75]
weight_list = [38, 55, 63, 70, 74, 78, 76, 82, 79, 73]
```

```python
plt.plot(np_age_list,np_weight_list,"g")
plt.title("Numpy Age vs Weight")
plt.xlabel("Age")
plt.ylabel("Weight")
plt.show()
```
