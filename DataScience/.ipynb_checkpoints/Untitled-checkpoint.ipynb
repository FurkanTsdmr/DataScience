```python
def salary_category(salary):
    if salary < 50000:
        return "Low"
    elif 50000 <= salary < 80000:
        return "medium"
    else:
        return "High"
```

```python
pd["Salary"].apply(salary_category)
```

```python
pd["Salary_Category"] = pd["Salary"].apply(salary_category)
```

```python
def new_Score(Experience,performance_Score):
    if Experience > 10:
       return performance_Score + 1
        
    else:
        return "Experince not greater 10 years"
```

```python
pd["YeniPuan"] = pd["Experience"].apply(new_Score)
```
