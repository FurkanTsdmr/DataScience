```python
installment_features = [
    column for column in df.columns
    if column.startswith("instal_")
]

print("Taksit feature sayısı:", len(installment_features))
print(installment_features)
```

```python
cc_missing = pd.DataFrame({
    "Eksik Sayısı": df[credit_card_features].isnull().sum(),
    "Eksik Yüzdesi": df[credit_card_features].isnull().mean() * 100
})

cc_missing.sort_values(
    "Eksik Yüzdesi",
    ascending=False
)
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    x=cc_missing["Eksik Yüzdesi"],
    y=cc_missing.index
)

plt.title("Kredi Kartı Feature'larında Eksik Veri Oranı")
plt.xlabel("Eksik Veri (%)")
plt.ylabel("Feature")

plt.show()
```

```python
#targert ilişkisi
cc_corr = (
    df[credit_card_features + ["TARGET"]]
    .corr(numeric_only=True)["TARGET"]
    .drop("TARGET")
    .sort_values(key=abs, ascending=False)
)

cc_corr
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    x=cc_corr.abs().values,
    y=cc_corr.abs().index
)

plt.title("TARGET ile Kredi Kartı Feature İlişkileri")
plt.xlabel("Absolute Correlation")
plt.ylabel("Credit Card Feature")

plt.show()
```
