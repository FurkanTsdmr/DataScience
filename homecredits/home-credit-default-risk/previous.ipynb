```python
df_copy = df_copy.drop(columns=["RATE_INTEREST_PRIMARY","RATE_INTEREST_PRIVILEGED"])
```

```python
df_copy["AMT_DOWN_PAYMENT"] = df_copy["AMT_DOWN_PAYMENT"].fillna(0)
df_copy["RATE_DOWN_PAYMENT"] = df_copy["RATE_DOWN_PAYMENT"].fillna(0)
```

```python
df_copy["AMT_CREDIT"] = df_copy["AMT_CREDIT"].fillna(df_copy["AMT_CREDIT"].median())
```

```python
df_copy["AMT_ANNUITY"] = df_copy["AMT_ANNUITY"].fillna(0)
df_copy["CNT_PAYMENT"] = df_copy["CNT_PAYMENT"].fillna(0)
```

```python
df_copy['AMT_GOODS_PRICE'] = df_copy['AMT_GOODS_PRICE'].fillna(df_copy["AMT_APPLICATION"])
```

```python
df_copy.isna().sum()
```

```python
df_copy["NAME_TYPE_SUITE"] = df_copy["NAME_TYPE_SUITE"].fillna("Unspecified")
```

```python
df_copy['PREV_IS_APPROVED'] = df_copy['DAYS_FIRST_DUE'].notna().astype(int)
```

```python
days_cols = [
    'DAYS_FIRST_DRAWING', 
    'DAYS_FIRST_DUE', 
    'DAYS_LAST_DUE_1ST_VERSION', 
    'DAYS_LAST_DUE', 
    'DAYS_TERMINATION'
]

for col in days_cols:
    df_copy[col] = df_copy[col].replace(365243, np.nan).fillna(0)

df_copy['NFLAG_INSURED_ON_APPROVAL'] = df_copy['NFLAG_INSURED_ON_APPROVAL'].fillna(0)
```

```python
print("Kalan eksik değer sayısı:", df_copy[days_cols + ['NFLAG_INSURED_ON_APPROVAL']].isna().sum().sum())
```

```python
import pandas as pd

# Müşteri bazlı (SK_ID_CURR) özet tablo oluşturma
df_client = (
    df.groupby("SK_ID_CURR")
    .agg(
        # Toplam başvuru sayısı
        TOTAL_PREV_APPLICATIONS=("SK_ID_PREV", "count"),
        # Kredi miktarlarının ortalaması ve toplamı
        AVG_AMT_CREDIT=("AMT_CREDIT", "mean"),
        TOTAL_AMT_CREDIT=("AMT_CREDIT", "sum"),
        # Ödeme miktarının ortalaması
        AVG_AMT_ANNUITY=("AMT_ANNUITY", "mean"),
        # Peşinatların ortalaması (Az önce 0 ile doldurmuştuk)
        AVG_AMT_DOWN_PAYMENT=("AMT_DOWN_PAYMENT", "mean"),
    )
    .reset_index()
)

# Yeni müşteri bazlı tablonun ilk hallerini kontrol edelim
print(df_client.head())
```
