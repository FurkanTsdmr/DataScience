```python
def argsSum(*args):
    return sum(args)
```

```python
#kwargs
def kwa(**kwargs):
    print(kwargs)
```

```python
kwa(appel=100,banana=150,melon=200)
```

```python
def kwa1(**kwargs):
    if "appel" in kwargs:
        print("applee")
    else:
        print(":(")
```
