# Special Methods

```python
class Fruit():

    def __init__(self,name,calories):
        self.name = name
        self.calories = calories

    def __str__(self):
        return f"{self.name}: {self.calories} caloires"

    def __len__(self):
        return self.calories
```

```python
myFruit = Fruit("banana",150)
```

```python
myFruit.calories
```

```python
myFruit.name
```

```python
print(myFruit)
```

```python
len(myFruit)
```

```python
class Train():

    def __init__(self,name):
        self.name = name
    def __getitem__(self,key):
        if key == "a":
            return self.name
        else:
            return "Not found"
```

```python
myTrain = Train("mytrain")
```

```python
myTrain["a"]
```

```python
myTrain["b"]
```

```python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"{self.name} - {self.age} yaşında"


student = Student("Ahmet", 21)
```
