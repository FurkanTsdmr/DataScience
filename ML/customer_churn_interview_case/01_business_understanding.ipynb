```python
#Veri Türlerini Ayrı Ayrı İnceleme
numeric_columns = df.select_dtypes(
    include=["int64","float"]
).columns.tolist()
categorical_columns = df.select_dtypes(
    include=["object","category"]
).columns.tolist()

len_numeric = len(numeric_columns)
len_cate = len(categorical_columns)


print(" Sayısal Değişkenler:",numeric_columns,"Toplam Col:",len_numeric, "\n")
print("Kategorik Değişkenler:",categorical_columns,"Toplam Kate:",len_cate,"\n")
```

```python
missing_count = df.isna().sum()

missing_percent = (df.isna().mean() * 100)

missing_report = pd.DataFrame({
    "missing_count":missing_count,
    "missing_percent":missing_percent
})

missing_report = missing_report.sort_values(
    by="missing_percent",
    ascending=False
)

display(missing_report)
```
