```python
y_pred_test = regression.predict(X_test)
```

```python
y_test
```

```python
#kıyaslama
plt.scatter(y_pred_test,y_test)
plt.show()
```

```python
MAE — Mean Absolute Error

İlk metriğimiz:

Mean Absolute Error

Yani:

Ortalama Mutlak Hata

Mantığı:

Gerçek - Tahmin

farkının mutlak değerini alıp ortalamasını bulmak.

Basitçe:

MAE = ortalama |y - ŷ|

şeklinde düşünebiliriz.

Örneğin:

Gerçek → 80
Tahmin → 75

hata: 5 olur.

MAE'nin güzel tarafı yorumlamasının kolay olmasıdır.

Eğer:

MAE = 4 ise modelimiz ortalama olarak yaklaşık 4 puanlık mutlak hata yapıyor diyebiliriz.
```

```python
MSE — Mean Squared Error

Daha önce teorik bölümde gördüğümüz:

Mean Squared Error

burada da karşımıza çıkıyor.

mse = mean_squared_error(y_test, y_pred_test)

MSE:

MSE = ortalama (y - ŷ)²

mantığıyla çalışır.

Hataların karesini aldığı için büyük hataları daha fazla cezalandırır.

Örneğin:

Hata = 2
→ 4

Hata = 10
→ 100

Bu nedenle büyük hatalar MSE üzerinde daha fazla etkiye sahip olur.

RMSE — Root Mean Squared Error

MSE'nin karekökünü alırsak:

rmse = np.sqrt(mse)

elde ederiz.

Bu:

Root Mean Squared Error

olarak adlandırılır.

Yani:

RMSE = √MSE

RMSE'nin önemli avantajlarından biri, hedef değişkenle aynı birimde olmasıdır.

Bizim örneğimizde hedefimiz:

Exam Score olduğu için RMSE'yi sınav puanı üzerinden yorumlayabiliriz.

R² Score

Bir diğer önemli metrik:

r2 = r2_score(y_test, y_pred_test)

Burada:

R² — R-squared değerini hesaplıyoruz.

R², modelin hedef değişkendeki varyasyonu ne ölçüde açıkladığını değerlendirmemize yardımcı olur.

Basit bir ifadeyle:

Modelimizin verideki değişimi ne kadar iyi açıkladığı hakkında fikir verir.

Örneğin R² değerinin 1'e yaklaşması, modelin gözlemlenen veriye daha iyi uyum sağladığını gösterebilir.

Ama burada da dikkatli olmak gerekiyor.

R²'nin yüksek olması modelin her durumda iyi olduğu anlamına gelmez.

Özellikle veri setinin yapısı, test yöntemi ve gerçek kullanım senaryosu birlikte değerlendirilmelidir.

Adjusted R²

Bir de:

Adjusted R² kavramı var.

Temel formülü:

1 - (1-r2) * (len(y_test)-1) / (
    len(y_test)-X_test.shape[1]-1
) şeklinde hesaplanabilir.

Adjusted R² özellikle birden fazla feature kullanan regresyon modellerinde önem kazanır.

Çünkü modele yeni değişkenler eklediğimizde normal R² bazı durumlarda artabilir.

Adjusted R² ise feature sayısını da hesaba katar.

Bizim şu an yalnızca:

Study Hours şeklinde tek bir feature'ımız olduğu için burada asıl önemli nokta formülü ezberlemekten ziyade kavramı tanımaktır.

Çoklu doğrusal regresyona geçtiğimizde bu metriğin neden daha anlamlı hale geldiğini daha net göreceğiz.
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
```

```python
mse = mean_squared_error(y_test,y_pred_test)
mae = mean_absolute_error(y_test,y_pred_test)
rmse = np.sqrt(mse)

print("mse",mse)
print("mae",mae)
print("rmse",rmse)
```

```python
r2 = r2_score(y_test,y_pred_test)
print("r2 score",r2)
```

```python
Modelimizi Birkaç Farklı Metrikle Ölçelim

Kodumuzu birlikte çalıştırabiliriz:

mse = mean_squared_error(y_test, y_pred_test)

mae = mean_absolute_error(y_test, y_pred_test)

rmse = np.sqrt(mse)

print("MSE:", mse)
print("MAE:", mae)
print("RMSE:", rmse)

Ardından:

r2 = r2_score(y_test, y_pred_test)

print("R2 Score:", r2)

ve Adjusted R²:

adjusted_r2 = 1 - (
    (1 - r2) *
    (len(y_test) - 1) /
    (len(y_test) - X_test.shape[1] - 1)
)

print("Adjusted R2:", adjusted_r2)

Artık modelimizi sadece:

"Grafikte güzel görünüyor."

diyerek değerlendirmiyoruz.

Elimizde ölçebileceğimiz metrikler var.
```

```python
#adjusted r2 score
1 - (1-r2)*(len(y_test)-1)/(len(y_test)-X_test.shape[1]-1)
```
