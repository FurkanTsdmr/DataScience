```python
sns.scatterplot(x=df["vehicle_age"],y=df["selling_price"],hue=df["fuel_type"])
plt.show()
#3-4 tane outlier var ciddi manada bozan bunları silebiliriz.
```

```python
df["selling_price"].max()
#medyan ve min e baktıgımızda bu 39milyon 500bin çok ucuk bir değer.
```

```python
#15milyon gibi bir sınıf çıkıp üstündekiler isilebiliri.z
df = df[df["selling_price"] < 15000000]
```

```python
df.describe()
#2 tane datadan kurtıulduk.
```

```python
sns.scatterplot(x=df["vehicle_age"],y=df["selling_price"],hue=df["fuel_type"])
plt.show()
#hatta daha temız olamsı için 10milyon dahi alabilirdik
#hatta 10 milyon alalım daha temiz olsun
```

```python
df=df[(df["selling_price"] <= 10000000)]
```

```python
sns.scatterplot(x=df["vehicle_age"],y=df["selling_price"],hue=df["fuel_type"])
plt.show()
```

```python
sns.boxplot(data=df,
             x="fuel_type",
             y="selling_price")
plt.show()
```

```python
sns.boxplot(data=df,
             x="transmission_type",
             y="selling_price")
plt.show()
```
