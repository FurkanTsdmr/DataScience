```python

sns.histplot(
    data = df,
    x="Exam Score",
    bins=30,
    kde=True
)
plt.show()
```

X Ekseni (Exam Score) : Öğrencilerin veya gözlemlerin aldığı sınav puanlarını gösterir (40 ile 100 arasınfda) <br>
Y Ekseni(Count) : Belirlenen puan aralıklarında kaç adet veri (öğrenci/gözlem) bulunduğunu gösterir. <br>
<hr>
Dağılım özellikleri <br>
1.1Sola Çarpıklık / Yüksek Puan Yoğunluğu : Grafiğe bakınca verilerin büyük kısmı 75- 100 puan aralığında kümelenmiş.<br>
1.2.En yüksek frekanslar (tepe noktası / mod) 78-80 puan aralığında ve bu aralıkta 4 gözlem bulunmaktadır.<br>
<hr>
Düşük Puanlar:<br>
2.1: 40 - 65 puan arasında daha az sayıda gözlem vardır ve dağılım seyrektir.<br>
2.2:Örneğin 42 - 44 puan ve 64 - 66 puan civarında 3'er gözlem yer almaktadır.<br>
<hr>
Mavi Çizgi (KDE EĞRİSİ)<br>
3.1: Yoğunluk eğirisi, puanlar yükseldikçe (özellikle 70'ten sonra) yükselen ve 80 - 90 bandında zirve yapıp 100'er doğru hafifçe alçalan bir eğilim gösterir. Bu durum, sınav sonuçlarının genel olarak yüksek puanlara doğru kaydığını(sola çarpık / negatively skewed bir dağılım sergilediğini) gösterir

#### Model Seçmeden önce tahmin etmek istediğimiz hedefi tanımlamalıyız.
##### Regresyon (Regression)
* Tahmin edilecek değer sayısal ve sürekli olduğunda kullanılır.
* Örnek: ev fiyatı,satış geliri,sıcaklık,ürün uzunluğu
* Modeller : Linear Regression,Ridge,Lasso,Random Forest Regressor,Gradient Boosting Regressot
##### Sımıflandırma (Classification)
* Tahmin edilecek sonuç bir sınıf veya kategoriyse kullanılır.
* Örnek:Sahtekarlık var/yok,Müşteri ayrılır/ayrılmaz,tür tahmini
* Modeller: Logistic Regression, Decision Tree Classifier, Random Forest Classifier, SVM, KNN, Naive Bayes.
##### Kümeleme (Clustering)
* Önceden verilmiş hedef sınıflar olmadan benzer gözlemleri gruplamak için kullanılır.
* Örnek:Müşteri segmentasyonu
* Modeller: K-Means, DBSCAN, Hierarchical Clustering.
##### Zaman Serisi Tahmini
* Zamana bağlı gözlemlerden gelecek değerleri tahmin etmek için kullanılır.
* Örnek:Günlük satış,talep,stok ihtiyacı.
* Yöntemler:Naive Forecast, Exponential Smoothing, ARIMA ve uygun makine öğrenmesi modelleri.

Bir de boyut indirgeme, anomali tespiti, öneri sistemleri, doğal dil işleme ve derin öğrenme gibi özel problem alanları vardır. Bunları ilerleyen bölümlerde ele alacağız.



#### Gözetimli Öğrenme
Gözetimli öğrenmede veriyi iki parçaya ayırırız.

```python
#Modelin tahmin yapmak için kullanacağı özellikler.
X = df.drop(columns=["Study Hours"],axis=1)
#Modelin öğrenmesini isteediğimiz hedef.
y = df["Exam Score"]
```

Bir hedefi tahmin ederken hedefin kendisini veya hedefi doğrudan ele veren başka bir sütunu X içinde bırakamayız. Buna target leakage(hedef sızıntısı) denir.
Kimlik numarası gibi sütunlar da yalnızca tanımlayıcı oldukları için çoğu problemde özellik olarak kullanılmamalıdır; fakat bazı durumlarda kimlik veya grup bilgisi veri yapısı hakkında anlamlı bilgiler taşıyabiliyor. Karar vermeden önce sütunu incele

```python
#Modelimizin eğitim,doğrulama ve test verileri için bu kütüphaneyi kullanırız.
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.20,random_state=15)
```

##### X,y -> Özellikler ve hedef
##### test_size=0.20 -> Verinin %20'si test için ayrılır.(80-20).
##### random_state=15 -> Bölmenin tekrarlanabilir olmasını sağlar.

#### Eksik Değerler ve Kategorik Veriler
##### Eksik Değer Doldurma


```python
from sklearn.impute import SimpleImputer
median_imputer = SimpleImputer(strategy="median")
mean_imputer = SimpleImputer(strategy="mean")
mode_imputer = SimpleImputer(strategy="most_frequent")
constant_imputer = SimpleImputer(strategy="constant",fill_value=0)
```

strategy seçenekleri:

"mean": Ortalama ile doldurur; uç değerlere duyarlıdır.
"median": Medyan ile doldurur; çarpık dağılımlarda daha dayanıklı olabilir.
"most_frequent": En sık görülen değer ile doldurur.
"constant": Belirlenen sabit değer ile doldurur.

Sabit 0 her zaman uygun değildir. Sıfır gerçek bir ölçümü temsil ediyorsa, eksik değerle sıfırı birbirine karıştırabilirsin.



##### One-Hot Encoding
Sıralı olmayan kategorileri ayrı sütunlara dönüştürür

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
encoder = OneHotEncoder(
    handle_unknown="ignore",
    drop="first",
    sparse_output=True
)
```

handle_unknown="ignore": Eğitim sırasında görülmeyen kategoriyle karşılaşınca dönüşüm hatasını önler.

drop="first": İsteğe bağlı olarak ilk kategoriyi çıkarır; bazı doğrusal modellerde çoklu doğrusal bağlantıyı azaltmak için kullanılabilir.

sparse_output=True: Varsayılan olarak seyrek matris üretebilir; çok sayıda kategori için bellek tasarrufu sağlar.

#### Ordinal Encoding
##### Kategoriler arasında gerçekten bir sıra varsa kullanılabilir

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
encoder = OrdinalEncoder(handle_unknown="use_encoded_value",unknown_value=-1)
```

Ordinal encoding kategorileri sayısal kodlara dönüştürür. Ancak kategorilere sayı vermek, her zaman aralarındaki mesafelerin anlamlı olduğu anlamına gelmez.

Örneğin <b>küçük, orta, büyük</b> doğal bir sıra taşır; <b>İstanbul, Ankara, İzmir</b> ise doğal bir büyüklük sırası taşımaz. İkinci tür kategoriler için One-Hot Encoding genellikle daha uygun bir başlangıçtır.
