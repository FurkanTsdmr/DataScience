```python
#feature : modelin tahmin yaparken kullandıgı giriş değişkenidir
#bu projede area,rooms,buldingage evin özelliklerini temsil eder.
#target ise modelin tahmin etmeye calstıgı değişkendir.
#Bu projede PRİCE: m2,oda sayısı,bina yasını vereceğiz model de fayatı tahmin etmeye çalışacak.
```

X : Giriş değişkenlerini <br>

y : Hedef değişkenlerini temsil eder.

```python
X = df.drop("Price",axis=1)
y=df["Price"]
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.20,random_state=42)
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
y_train.shape
```

```python
y_test.shape
```

```python
model = LinearRegression()
```

```python
model.fit(X_train,y_train)
```

```python
model.coef_
```

```python
coefficients = pd.DataFrame({
    "Feature": X_train.columns,
    "Coefficient": model.coef_
})

coefficients
```

```python
model.intercept_
```

```python
y_pred = model.predict(X_test)
```

```python
comprasion = pd.DataFrame({
    "Actual Price" : y_test.values,
    "Predicted Price" : y_pred
})
comprasion.head(10)
```

```python
mae = mean_absolute_error(y_test,y_pred)
print("MAE:",mae )
```
