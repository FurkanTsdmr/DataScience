```python
class Student:
    def __init__(self, name, age, department):
        self.name = name
        self.age = age
        self.department = department
    def introduce(self):
        print(f"Benim adım {:self.name}")
        print(f"Benim yaşım: {self.age}")
```

```python
student3 = Student("Veli",24,"Data Science")
student3.introduce()
```

```python
student1 = Student("Ahmet",21,"Data")
```

```python
student2 = Student("mehmet",23,"Computer Science")
```

```python
print(student1.name)
print(student1.age) 
print(student1.department)
```

```python
class Musician():

    def __init__(self,name):
        self.name=name
        print("musician class")

    def test1(self):
        print("test1")

    def test2(self):
        print("test2")
```

```python
furkan = Musician("furkan")
```

```python
furkan.name
```

```python
furkan.test2()
```

## polymorphism:aynı isimle farklı işlem yapma

```python
class Banana():

    def __init__(self,name):
        self.name = name

    def info(self):
        return f"100 calorie {self.name}"
        
class Apple():
    def __init__(self,name):
        self.name = name

    def info(self):
        return f"150 calorie {self.name}"
```

```python
banana = Banana("banana")
```

```python
apple = Apple("apple")
```

```python
banana.info()
```

```python
apple.info()
```

```python
fruitList = [banana,apple]
```

```python
for fruit in fruitList:
    print(fruit.info())
```

## Encapsulation : hapsetmek demektir. 
Oluşturulan değişkenlerin nereden erişebildiğidir

```python
class Phone():

    def __init__(self,name,price):
        self.name = name
        self.__price = price

    def info(self):
        print(f"{self.name} price is : {self.__price}")

    def changePrice(self,price):
        self.__price = price
```

```python
iphone = Phone("iphone 17",700)
```

```python
iphone.info()
```

```python
iphone.__price = 400
```

```python
iphone.info()
```

```python
iphone.changePrice(300)
```

```python
iphone.info()
```

# Abstraction : soyutlama:
Soyut bit sınıf olusturuyoruz. Normal sınıflar gibi değil
bize bir plan yapsın diye kullanıyoruz.
Mimar çizer falan tüm daireler o plana göre yapılır.
Kendisini kullnmayız yapısını kullanarak ilerleriz

```python
from abc import ABC,abstractmethod
```

```python
class Car(ABC):
    
    @abstractmethod
    def maxSpeed(self):
        pass
```

```python
class Tesla(Car):
    def maxSpeed(self):
        print("200km")
```

```python
tesla = Tesla()
```

```python
tesla.maxSpeed()
```

```python
class Mercedes(Car):

    def maxSpeed(self):
        print("170km")
```

```python
mercedes = Mercedes()
```

```python
mercedes.maxSpeed()
```
